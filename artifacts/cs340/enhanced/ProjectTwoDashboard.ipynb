```python
# Imports and dashboard setup
from IPython.display import display
from ProjectOne_CRUD_Python_Module import AnimalShelter
from dashboard_helpers import (
    build_breed_distribution_figure,
    build_rescue_dataframes,
    display_logo,
    records_to_dataframe,
    summarize_map_focus,

)

# Security note: MongoDB credentials are loaded from environment variables
# instead of being hardcoded in the notebook source.
db = AnimalShelter()
display_logo("Grazioso Salvare Logo.png")

preview_fields = {
    "name": 1,
    "breed": 1,
    "animal_type": 1,
    "age_upon_outcome_in_weeks": 1,
    "outcome_type": 1,
}
dashboard_fields = {
    "name": 1,
    "breed": 1,
    "location_lat": 1,
    "location_long": 1,
}

preview_df = records_to_dataframe(
    db,
    projection=preview_fields,
    limit=25,
    )
rescue_frames = build_rescue_dataframes(db)
dashboard_df = records_to_dataframe(db, projection=dashboard_fields)

print(f"Preview rows loaded: {len(preview_df)}")
preview_df.head()
```

```python
# Rescue filter outputs
for rescue_name, dataframe in rescue_frames.items():
    print(f"{rescue_name.title()} rescue matches: {len(dataframe)}")
    display(dataframe.head())

# Visualization output
breed_figure = build_breed_distribution_figure(dashboard_df)
if breed_figure is not None:
    breed_figure.show()
else:
    print("Breed distribution chart could not be generated.")

print(summarize_map_focus(dashboard_df))

# Validation example for the enhanced CRUD layer
sample_document = {
    "animal_type": "Dog",
    "breed": "Labrador Retriever Mix",
    "color": "Black",
    "date_of_birth": "2021-01-01",
    "outcome_type": "Adoption",
    "sex_upon_outcome": "Neutered Male",
    "age_upon_outcome_in_weeks": 52,
    "location_lat": 30.75,
    "location_long": -97.48,
}
is_valid, validation_message = db.validate_document(sample_document)
print(f"Validation example: {is_valid} - {validation_message}")
```

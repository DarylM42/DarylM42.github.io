```python
# Imports
import pandas as pd
import plotly.express as px
from ProjectOne_CRUD_Python_Module import AnimalShelter
from IPython.display import Image, display

# Connect to MongoDB
username = "aacuser"
password = "zoey0728"
db = AnimalShelter(username, password)

# Display Grazioso Salvare logo
display(Image(filename="Grazioso Salvare Logo.png"))

# Data preview
df = pd.DataFrame.from_records(db.read({}))
if "_id" in df.columns:
    df.drop(columns=["_id"], inplace=True)
df.head()

# Water rescue filter
water_query = {
    "$and": [
        {"age_upon_outcome_in_weeks": {"$lte": 104}},
        {"breed": {"$in": ["Labrador Retriever Mix", "Chesapeake Bay Retriever", "Newfoundland"]}}
    ]
}
df_water = pd.DataFrame.from_records(db.read(water_query))
if "_id" in df_water.columns:
    df_water.drop(columns=["_id"], inplace=True)
df_water.head()

# Mountain/wilderness rescue filter
mountain_query = {
    "$and": [
        {"age_upon_outcome_in_weeks": {"$lte": 104}},
        {"breed": {"$in": ["German Shepherd", "Border Collie", "Bloodhound"]}}
    ]
}
df_mountain = pd.DataFrame.from_records(db.read(mountain_query))
if "_id" in df_mountain.columns:
    df_mountain.drop(columns=["_id"], inplace=True)
df_mountain.head()

# Disaster/tracking filter
disaster_query = {
    "$and": [
        {"age_upon_outcome_in_weeks": {"$lte": 104}},
        {"breed": {"$in": ["Doberman Pinscher", "Rottweiler", "Belgian Malinois"]}}
    ]
}
df_disaster = pd.DataFrame.from_records(db.read(disaster_query))
if "_id" in df_disaster.columns:
    df_disaster.drop(columns=["_id"], inplace=True)
df_disaster.head()

# Reset filter (full dataset)
df_reset = pd.DataFrame.from_records(db.read({}))
if "_id" in df_reset.columns:
    df_reset.drop(columns=["_id"], inplace=True)
df_reset.head()

# Breed distribution chart
fig = px.pie(df_reset, names="breed", title="Breed distribution")
fig.show()

# Map information (simulated)
lat = df_reset.iloc[0]["location_lat"]
lon = df_reset.iloc[0]["location_long"]
breed = df_reset.iloc[0].get("breed", "Unknown")
name = df_reset.iloc[0].get("name", "Unknown")
print(f"Map would center on: {lat}, {lon} — {name} ({breed})")
```
